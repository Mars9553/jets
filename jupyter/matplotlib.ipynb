```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
r = np.arange(12, 42)
r
```

```python
y = r * 3.482
y
```

```python
plt.title('Linear Graph')
plt.xlabel('x-point')
plt.ylabel('y-point')
plt.plot(r, y, color= 'r', linestyle= ':')
plt.show()
```

```python
a = np.arange(11, 53)
b1 = a * 3
b2 = a * 2.383
```

```python
plt.title('Linear Graph')
plt.xlabel('x-point')
plt.ylabel('y-point')
plt.plot(b1, color= 'r', linestyle= ':', linewidth = 2)
plt.plot(b2, color= 'b', linestyle= ':', linewidth = 2.5)
plt.grid(True)
plt.show()
```

```python
plt.subplot(1,2,2)
plt.plot(b1, color= 'r', linestyle= ':', linewidth = 2)
plt.grid(True)

plt.subplot(1,2,1)
plt.plot(b2, color= 'b', linestyle= ':', linewidth = 2.5)

plt.title('Linear Graph')
plt.xlabel('x-point')
plt.ylabel('y-point')
plt.grid(True)
plt.show()
```

```python
algo = {'Rio':82, 'Kai':63, 'Mia':99, 'Sam':71}
```

```python
a = list(algo.values())
b = list(algo.keys())
```

```python
plt.bar(b, a, color= 'purple', linewidth = 2.5)
plt.title('Bar Graph')
plt.grid()
plt.ylabel('Student Scores')
plt.xlabel('Student Name')
```

```python
plt.barh(b, a, color= 'purple')
plt.title('Bar Graph')
plt.grid()
plt.ylabel('Student Scores')
plt.xlabel('Student Name')
```
